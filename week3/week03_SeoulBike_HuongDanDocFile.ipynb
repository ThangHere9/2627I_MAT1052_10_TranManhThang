````python
import pandas as pd
````

Notebook luôn làm việc trong một **thư mục hiện tại** (*current working directory*).

## 3. Đọc file CSV

Hàm `pd.read_csv()` dùng để đọc dữ liệu từ file CSV.

Cú pháp cơ bản:

```python
pd.read_csv("ten_file.csv")
```

Với bộ dữ liệu này:

```python
pd.read_csv("SeoulBikeData.csv", encoding="cp949")
```

Trong đó:

- `"SeoulBikeData.csv"` là tên file cần đọc;
- `encoding="cp949"` cho Python biết bảng mã ký tự của file gốc;
- kết quả được lưu vào biến `bike`.

`bike` lúc này là một **DataFrame**, tức một bảng dữ liệu trong pandas.


````python
bike = pd.read_csv(
    "SeoulBikeData.csv",
    encoding="cp949"
)
````

## 4. Kiểm tra dữ liệu sau khi đọc

Sau khi đọc file, không nên bắt đầu phân tích ngay. Trước hết cần kiểm tra xem dữ liệu đã được đọc đúng hay chưa.

### `bike.head()`

Hiển thị 5 hàng đầu tiên của bảng.


````python
bike.head()
````

### `bike.shape`

Cho biết kích thước của bảng dưới dạng:

```text
(số hàng, số cột)
```

Ví dụ `(8760, 14)` nghĩa là có 8.760 hàng và 14 cột.


````python
bike.shape
````

### `bike.columns`

Hiển thị tên tất cả các cột trong dữ liệu.


````python
bike.columns
````

````python
bike.columns = [
    "date", "rentals", "hour", "temp", "humidity",
    "wind", "visibility", "dewpoint", "solar",
    "rain", "snow", "season", "holiday", "functioning"
]
bike.head()
````

## 5. Nếu gặp lỗi

### `FileNotFoundError`

Nếu xuất hiện lỗi dạng:

```text
FileNotFoundError
```

thì Python không tìm thấy `SeoulBikeData.csv`.

Hãy kiểm tra:

1. file CSV và notebook có nằm cùng thư mục hay không;
2. tên file có đúng là `SeoulBikeData.csv` hay không;
3. chạy lại:

```python
Path.cwd()
Path("SeoulBikeData.csv").exists()
```

### `UnicodeDecodeError`

Với file Seoul Bike gốc, cần giữ:

```python
encoding="cp949"
```

Nếu bỏ phần này, Python có thể không đọc đúng bảng mã ký tự.
